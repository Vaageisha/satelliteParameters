# PART 1 : Understanding a FITS File (2D Image)




```python
# Step 1: Open the FITS file
```

```python
from astropy.io import fits

fits_file = "image.fits"  # the location of your file will go here

with fits.open(fits_file) as hdul:
    hdul.info()
```

```python
'''
Typical output:
No.    Name      Type      Dimensions
0      PRIMARY   PrimaryHDU   (n, n)

This tells us:

Number of HDUs
Type of HDU
Dimensions of the data
'''
```

```python

# Step 2: Read the Header
# The header contains metadata describing the image.
```

```python
from astropy.io import fits

fits_file = "image.fits"  #the location of your file will go here

with fits.open(fits_file) as hdul:

    header = hdul[0].header

    print(header)
```

```python
'''
Some useful keywords:

Keyword	  Meaning

NAXIS	    Number of dimensions
NAXIS1	  Width
NAXIS2	  Height
BITPIX	  Data type
BSCALE	  Scaling factor
BZERO	    Offset
EXPTIME	  Exposure time
DATE-OBS	Observation time

'''
```

```python
print("BITPIX =", header["BITPIX"])
print("NAXIS =", header["NAXIS"])
print("NAXIS1 =", header["NAXIS1"])
print("NAXIS2 =", header["NAXIS2"])
```

```python
# Step 3: Understand BITPIX

'''
Common values:

BITPIX	Meaning

8	      Unsigned integer
16	    16-bit integer
32	    32-bit integer
-32	    Float32
-64	    Float64

Example:

BITPIX = 16
means pixel values are stored as 16-bit integers.

'''
```

```python
# Step 4: Load the Image Data
```

```python
with fits.open(fits_file) as hdul:

    image = hdul[0].data

print(image.shape)
print(image.dtype)
```

```python
'''
Example:

(1024, 1024)  [dimensions of the fits file]
uint16

'''
```

```python
# Step 5: Display the Image
```

```python
import matplotlib.pyplot as plt

plt.imshow(image, origin="lower", cmap="gray")
plt.colorbar()
plt.show()
```

```python
# Step 6: Basic Image Statistics

# This gives the first idea of signal and noise.
```

```python
import numpy as np

print("Minimum =", np.min(image))
print("Maximum =", np.max(image))
print("Mean =", np.mean(image))
print("Std =", np.std(image))
```





#  PART 2: Working with a FITS Cube (3D Data)

```python
'''

A cube contains many frames.
Instead of: (height, width)

we have:

(frames, height, width)

Example:

(500, 1024, 1024)

means:

500 frames
each frame is 1024 × 1024 pixels
'''
```

```python
# Step 1: Open the Cube
```

```python
from astropy.io import fits

fits_file = "cube.fits"

with fits.open(fits_file) as hdul:

    cube = hdul[0].data

print(cube.shape)
```

```python
# Step 2: Extract a Frame
```

```python
frame_number = 377
frame = cube[frame_number]
print(frame.shape)
```

```python
'''
gives
(1024, 1024)
which is a normal 2D image.
'''
```

```python
# Step 3: View the Frame
```

```python
import matplotlib.pyplot as plt

plt.imshow(frame, origin="lower", cmap="gray")
plt.colorbar()
plt.show()
```

```python
# Step 4: Save the Frame as a FITS File
```

```python
from astropy.io import fits

fits.writeto(
    "frame_377.fits",
    frame,
    overwrite=True
)
```

```python
# Step 5: Save as PNG
```

```python
import matplotlib.pyplot as plt

plt.imsave(
    "frame_377.png",
    frame,
    cmap="gray",
    origin="lower"
)
```

```python
'''
Part 3: Understanding the Noise Model

Once we know how to read FITS files and cubes, we can ask:

Why are pixel values fluctuating even when nothing is happening?

The measured pixel value is typically:
Signal=Object Signal+Sky Background+Dark Current+Read Noise

For CCD/CMOS astronomy images:
σtotal​= (σshot2​+σdark2​+σread2​) ^ 1/2
​	​

where:

Shot noise comes from photons
Dark noise comes from thermal electrons
Read noise comes from electronics



'''
```

```python
'''
Measuring Noise from Real Data
Select a blank sky region:
'''
sky_region = image[100:200, 100:200]
```

```python
# The standard deviation is the observed noise level.

import numpy as np

mean = np.mean(sky_region)
std = np.std(sky_region)

print(mean)
print(std)
```
